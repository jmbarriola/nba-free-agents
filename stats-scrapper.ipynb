```python
# pip install lxml
import requests
from bs4 import BeautifulSoup
from io import StringIO
import pandas as pd
a = []



for n in range(0, 45*25, 25):
    if n%100 == 0:
        print(n)
    url = f"https://www.prosportstransactions.com/basketball/Search/SearchResults.php?Player=&Team=&BeginDate=2014-05-01&EndDate=2014-12-31&PlayerMovementChkBx=yes&Submit=Search&start={n}"

    response = requests.get(url)
    html = response.content
    soup = BeautifulSoup(html, 'lxml')
    table = soup.find('table', attrs={'class': 'datatable center'})
    table_str = str(table)

    # Use StringIO to read the string as if it were a file
    df = pd.read_html(StringIO(table_str), header=0)[0]
    free_agents = df[df['Notes'].str.contains('agen')]
    a.append(free_agents)
```

```python
free_agents = pd.concat(a)
free_agents.Acquired = free_agents.Acquired.str.replace('•','') # remove • in front of player's name
free_agents.Relinquished = free_agents.Relinquished.str.replace('•','') # remove • in front of player's name
```

```python
import re

# Define a function to extract year and money
def extract_year_money(notes):
    match = re.search(r'signed .* (\d+)-year.*\$(\d.*M)', notes)
    if match:
        return pd.Series([int(match.group(1)), match.group(2)])
    else:
        return pd.Series([None, None])

# Apply the function to the Notes column
free_agents[['Contract_Years', 'Contract_Value']] = free_agents['Notes'].apply(extract_year_money)
```
