```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL for NBA free agents (change year as needed)
year = 2024  # Change this to the desired year
url = f"https://www.spotrac.com/nba/free-agents/{year}/"

# Headers to mimic a real browser request
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, "html.parser")

# Find the table containing free agents
table = soup.find("table", class_="datatable")
if not table:
    print("No table found on the page.")
    exit()

# Extract table headers
headers = [th.text.strip() for th in table.find_all("th")]

# Extract table rows
data = []
for row in table.find_all("tr")[1:]:  # Skip header row
    cols = row.find_all("td")
    if cols:
        data.append([col.text.strip() for col in cols])

# Create DataFrame and save to CSV
df = pd.DataFrame(data, columns=headers)
df.to_csv(f"nba_free_agents_{year}.csv", index=False)

print(f"NBA Free Agents {year} data saved to nba_free_agents_{year}.csv")
```

```python
class NBAFreeAgentScraper:
    def __init__(self, url):
        self.url = url
        self.headers = []
        self.data = []

    def scrape_data(self):
        response = requests.get(self.url)
        response.encoding = 'utf-8'
        soup = BeautifulSoup(response.text, 'html.parser')
        table = soup.find('table')
        self.headers = table.find('thead').find_all('th')
        self.headers = [header.text.strip() for header in self.headers]
        rows = table.find('tbody').find_all('tr')
        rows = [row.find_all('td') for row in rows]
        self.data = []
        for row in rows:
            player = [data.text.strip() for data in row]
            self.data.append(player)

    def save_to_csv(self, filename):
        with open(filename, 'w', newline='') as csvfile:
            csvwriter = csv.writer(csvfile)
            csvwriter.writerow(self.headers)
            csvwriter.writerows(self.data)

# Use the NBAFreeAgentScraper class to scrape data from the NBA website
url = 'https://www.spotrac.com/nba/free-agents/'
scraper = NBAFreeAgentScraper(url)
scraper.scrape_data()
scraper.save_to_csv('nba_free_agents.csv')

pd.read_csv('nba_free_agents.csv')
```
